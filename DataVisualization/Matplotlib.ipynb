```python
import matplotlib.pyplot as plt
import numpy as np
```

# Matplotlib Basic

```python
x = np.linspace(0,5,11)
y = x ** 2
x, y
```

```python
# create plot
plt.plot(x, y)

#give xlabel, ylabel, graph title
title = plt.title("Line Graph")
xlable = plt.xlabel("x-axis")
ylabel = plt.ylabel("y-axis")
```

# subplot in matplotlib

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(x,y)
plt.subplot(2,2,3)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(x,y)
```

# object oriented way
Matplotlib allows the aspect ratio, DPI and figure size to be specified when the Figure object is created. You can use the figsize and dpi keyword arguments.

figsize is a tuple of the width and height of the figure in inches dpi is the dots-per-inch (pixel per inch).

```python
fig = plt.figure(figsize=(12,8), dpi=100)
axis1 = fig.add_axes([0.1, 0.1, 0.4, 0.4])
axis1.plot(x,y)
```

```python
# if we give same for two axes it overlaps
fig = plt.figure(figsize=(10,10), dpi=60)
axis1 = fig.add_axes([0.1, 0.1, 1,1])
axis1.plot(x,y)
axis2 = fig.add_axes([0.3,0.4,0.3,0.3])
axis2.plot(x,y)
```

# types of plot


```python
#1.scatter plot
plt.scatter(x,y)
```

```python
#2.Histogram
plt.hist(x)
```

```python
#3.boxplot
plt.boxplot(x)
```

# Saving plot

```python
fig = plt.figure(figsize=(5,5), dpi=100)
axis = fig.add_axes([0.1, 0.1, 0.3,0.3])
axis.plot(x,y)
plt.savefig("basic.png")
```

# Working with Images

```python
import matplotlib.image as mpimg
img = mpimg.imread("basic.png")
plt.imshow(img)

#dont want axis
plt.axis('off')

#cropped image
cropimg = img[300:600, 10:120]
plt.imshow(cropimg)
```
