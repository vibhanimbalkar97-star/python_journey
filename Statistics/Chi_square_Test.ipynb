```python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency
import seaborn as sns
```

```python
#load data
df = sns.load_dataset('titanic')
df
```

```python
#Contingency table = the table itself
#pd.crosstab() = Pandas function used to create that table

contingency_table = pd.crosstab(df['pclass'], df['sex'])
contingency_table
```

```python
p_value, chi, df, expected = chi2_contingency(contingency_table)
p_value
```

```python
alpha = 0.05
if p_value > alpha:
    print("Accept Null hypothesis")
else:
    print("Reject Null Hypothesis")
```
