```python
import numpy as np
from scipy import stats
```

```python
sample = [172, 174, 168, 169, 171, 173, 175, 170, 169, 172]
sample_mean = np.mean(sample)
sample_std = np.std(sample,ddof = 1)
n = len(sample)
population_mean = 170
```

```python
t_stats = (sample_mean - population_mean ) / (sample_std/np.sqrt(n)) 
t_stats
```

```python
p_value = 2 * (1 - stats.t.cdf(abs(t_stats), df=n-1))
p_value
```

```python
alpha = 0.05
if p_value > alpha : 
    print("Accpt null hypothesis")
else:
    print("Reject null hypothesis")
```

A company claims the average employee productivity score is 70.
Calculate:

1.Sample mean
2.Sample standard deviation
3.T-score
4.p-value
5.Reject H₀ or Fail to reject H₀

```python
data = np.array([
    74, 68, 72, 71, 75,
    69, 73, 70, 76, 72
])

population_mean = 70
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)
n=len(data)
```

```python
t_stats = (sample_mean - population_mean) / (sample_std/np.sqrt(n))
t_stats
```

```python
p_value = 2 * (1 - stats.t.cdf(abs(t_stats), df=n-1))
p_value
```

```python
alpha = 0.05
if p_value > alpha:
    print("Accept Null hypothesis")
else:
    print("Reject Null Hypothesis")
```

# Independent Two-sample T-test
Find:We want to check whether students taught using Method A and Method B have different average scores.

Why equal_var=False?
It performs Welch's independent two-sample t-test, which does not assume both groups have the same variance.

1.Mean of A
2.Mean of B
3.T-score
4.p-value
5.Decision: Reject H₀ / Fail to reject H₀

```python
method_A = np.array([
    72, 75, 70, 68, 74,
    71, 73, 69, 76, 72
])

method_B = np.array([
    78, 82, 80, 85, 79,
    83, 81, 84, 77, 80
])
```

```python
mean_A = np.mean(method_A)
mean_B = np.mean(method_B)
mean_A
```

```python
t_score, p_value = stats.ttest_ind(method_A, method_B, equal_var=False)
p_value
```

```python
alpha = 0.05
if p_value > alpha:
    print("Accept Null hypothesis")
else:
    print("Reject Null Hypothesis")
```
