```python
import numpy as np
import pandas as pd
```

# Finding Missing Data = isna()

```python
data = {
    'A': [1, 2, np.nan, 4, 5],
    'B': [1, 2, 3, 4, 5],
    'C': [1, 2, 3, np.nan, np.nan],
    'D': [1, np.nan, np.nan, np.nan, 5]
}

d1 = pd.DataFrame(data)
d1
```

```python
d1.isna()
```

# count of missing or null values =isna().sum()= gives columwise count

```python
d1.isna().sum()
```

# tell all the columns have null value or not = isna().any()

```python
d1.isna().any()
```

# Removing Missing Data = dropna() = work row wise, all rows removed where null values are given.

```python
d1
```

```python
d1.dropna()
```

# thresh = value => whatever value is written at least that much not null value needed. If nit then row removed.

```python
d1.dropna(thresh = 3)
```

# Filling the missing Data = fillna(value) = column wise selection.

```python
d1
```

```python
d1.fillna(0)
```

# fill different values for each column

```python
values = {'A': 1, 'B': 2, 'C': 3, 'D': 4}
d1.fillna(value=values)
```

# fill mean value = fillna(d1.mean())

```python
d1.fillna(d1.mean())
```
