# Pandas Series
A Series is a one-dimensional labeled array capable of holding any data type. The axis labels are collectively called the index.

# Creating series
Multiple ways to create series = list, numpy array, dict

```python
import numpy as np
import pandas as pd
```

```python
labels=['a', 'b','c']
list = [10,20,30]
dict ={
    'a':10,
    'b':20,
    'c':30
}
array = np.array([10,20,30])
```

```python
pd.Series(list)
```

```python
#custom label creation
pd.Series(list, index=labels)
```

```python
pd.Series(array)
```

```python
pd.Series(dict)
```
