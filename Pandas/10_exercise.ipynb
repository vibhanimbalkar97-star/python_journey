# iloc vs loc
iloc → "Give me the data at this POSITION"
loc  → "Give me the data with this LABEL / CONDITION"

```python
data = {
    "Name": ["Amit", "Neha", "Rahul", "Priya", "Sneha", "Vikas"],
    "Age": [22, 19, 25, 21, 28, 18],
    "City": ["Pune", "Mumbai", "Pune", "Delhi", "Mumbai", "Pune"],
    "Salary": [30000, 25000, 45000, 35000, 50000, 22000]
}

df1 = pd.DataFrame(data)
df1
```
