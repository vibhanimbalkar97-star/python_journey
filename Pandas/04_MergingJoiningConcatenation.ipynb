# Merging 2 dataframes = need two data fo this. pd.merge(d1, d2)

```python
employees = pd.DataFrame({
    'employee_id': [1, 2, 3, 4, 5],
    'name': ['John', 'Anna', 'Peter', 'Linda', 'Bob'],
    'department': ['HR', 'IT', 'Finance', 'IT', 'HR']
})

# DataFrame 2: Salary information
salaries = pd.DataFrame({
    'employee_id': [1, 2, 3, 6, 7],
    'salary': [60000, 80000, 65000, 70000, 90000],
    'bonus': [5000, 10000, 7000, 8000, 12000]
})
```

```python
employees
```

```python
salaries
```

```python
pd.merge(employees,salaries)
```

```python
pd.merge(salaries,employees)
```

# on = take a common column between 2 DF
# how = inner ===== take only common parts from column
# how = outer ===== takes all values from both data, where no value take as NaN
# how = left ======take columns from d1, d2 and gives only values of d1 where no value NaN
# how = right ======take columns from d1, d2 and gives only values of d12 where no value NaN
# note = use inner to save from null values

```python
employees
```

```python
salaries
```

```python
pd.merge(employees, salaries, on='employee_id', how='inner')
```

```python
pd.merge(employees, salaries, on='employee_id', how='outer')
```

```python
pd.merge(employees, salaries, on='employee_id', how='left')
```

```python
pd.merge(employees, salaries, on='employee_id', how='right')
```

# Concatination of 2 dataframes = pd.concat([d1,d2])

```python
df1 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2'],
    'C': ['C0', 'C1', 'C2']
})

df2 = pd.DataFrame({
    'A': ['A3', 'A4', 'A5'],
    'B': ['B3', 'B4', 'B5'],
    'C': ['C3', 'C4', 'C5']
})
```

```python
df1
```

```python
df2
```

# by default column wise axis=0

```python
pd.concat([df1,df2])
```

# row wise concat = axis=1

```python
pd.concat([df1,df2], axis=1)
```

# Joining 2 data frames = d1.join(d2)

```python
df1 = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Charlie']
}, index=[1, 2, 3])

# Second DataFrame
df2 = pd.DataFrame({
    'score': [85, 90, 75]
}, index=[2, 3, 4])
```

```python
df1
```

```python
df2
```

```python
df1.join(df2)
```

```python
df2.join(df1)
```

```python
df1.join(df2, how='outer')
```
