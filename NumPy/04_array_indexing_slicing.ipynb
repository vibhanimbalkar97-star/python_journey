# access/slice single element = arr[index]
# slice the portion of vector = arr[start:end:step]


```python
import numpy as np
```

```python
numbers = np.arange(1, 11)
numbers
```

```python
#slice or access single element
single = numbers[8]
single
```

```python
#slice some portion
res = numbers[2:8]
res
```

```python
res = numbers[:6]
res
```

```python
res = numbers[:]
res
```

```python
res = numbers[2:]
res
```

# access single matrix = arr[row,column]
# slice column = arr[:column]
# slice row = arr[]

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
res = arr[1,2]
res
```

```python
res = arr[4]
res
```

```python
res = arr[:,3]
res
```

```python
res = arr[1:5,1:4]
res
```

```python
res = arr[2:,3:]
res
```

```python
res = numbers[2::3]
res
```
