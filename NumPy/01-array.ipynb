# Numpy
  same data type like only list, tuple, numbers, string. NumPy arrays are generally designed to contain one common data type (homogeneous data).
  But if you put different data types in the same NumPy array, NumPy will usually convert them to a common compatible type.

 NumPy arrays are homogeneous, meaning they normally store elements of the same data type. If we provide compatible mixed types, NumPy performs type promotion and converts them to a common dtype. If necessary, NumPy can also use object dtype for arbitrary Python objects, but that loses many of the performance advantages of normal NumPy arrays.

Numpy asks like = "What single data type can hold all three values without losing information?"
Integer → ❌ "30" cannot be an integer automatically
Float   → ❌ "30" is still a string
String  → ✅ can represent 10, 20 and "30"

int + float       → float
int + bool        → integer (typically)
int + string      → string
float + string    → string
